## Part 1

(Atomic units are used throughout this section.)

The main goal of this part of the assignment is to study the vibronic spectrum of the interhalogen compound ICl (which you have synthesised in year 2), specifically the band due to transitions from the X<sup>1</sup>&Sigma;<sup>+</sup> ground electronic state to the A<sup>3</sup>&Pi;<sub>1</sub> excited state. Although this is a transition from a singlet to a triplet state and therefore forbidden by the spin selection rules, it is a rather intense transition illustrating the importance of spin-orbit coupling in ICl.

We will consider the following potential energy curves of the two states involved.

![Potential energy curves for ICl](Part 1/ICl_curves.svg)

The potential energy curves for each state are represented by the Lippincott potential:
$$V(x)=D_e\left(1-e^{-\gamma^2\frac{r_e}{x}(x-r_e)^2}\right)\left(1-\frac{4}{5}\left(b \gamma-\frac{b^2}{r_e}\right)\sqrt{\frac{r_e}{x}}(x-r_e) e^{-b \gamma (x-r_e)}\right)+T_e$$

where b=1.065 is an empirical constant.

The potential parameters used for each state are:

||X <sup>1</sup>&Sigma;<sup>+</sup>|A<sup>3</sup>&Pi;<sub>1</sub>|
|----------------------------------|---------|---------|
|D<sub>e</sub>/E<sub>h</sub>       |0.079999 |0.017385 |
|r<sub>e</sub>/a<sub>0</sub>       |4.3858   |5.0877   |
|&gamma;/a<sub>0</sub><sup>-1</sup>|0.9790   |1.1602   |
|T<sub>e</sub>/E<sub>h</sub>       |0.10     |0.1626   |

We will consider that the wavefunction of the system in the ground state is the lowest eigenstate of the harmonic oscillator with a force constant consistent with the ground state potential $k=\left.\frac{d^2 V}{d x^2}\right|_{x=r_e}$, which in this case is $k=2 D_e \gamma^2$.

Although spin forbidden, we will treat the transition as being dipole allowed and consider that the following third order polynomial provided a good fit to the transition dipole moment &mu;<sub>XA</sub> between the two electronic states dependence as a function of the inter-atomic distance (in atomic units) around the ground state equilibrium distance:
$$\mu_{XA}(x)=-0.036 x^3 + 0.428 x^2 -1.64 x + 2.03$$

The accompanying file [qdyn.py](Part 1/qdyn.py) includes the Python functions `propagator()` and `animate_dynamics()`. The former implements the time-evolution operator for a system given an arbitrary potential,

$$\varphi(x,t_0+\Delta t)=e^{-i \hat{H} \Delta t}\varphi(x,t_0),$$

while the latter can be used to display the evolution of the wavefunction of the system with time. Refer to the functions' documentation for details on their use.

### Technical tips

* To make sure you have defined the functions for the potential correctly, start by plotting them to reproduce the figure above.
* Overlap the plot of your initial wavefunction and the transition dipole moment with your potential curves, to make sure these are defined correctly.
* In order to obtain valid results it is critical to set up an adequate grid of space points. The grid should be fine enough to describe high momentum components of the wavefunction. You should set the limits of the grid such that the wavefunction remains equal to zero at the limits of the grid throughout all the simulation to avoid artefacts. Check that this is the case before doing any further calculations.
* The time-step should be small enough to propagate the wavefunction correctly. But a too small time-step may require too many steps to be taken, and depending on how fine is your space grid this may need greater memory resources than those available in the computer you are using. Make tests with short runs, and simulate for longer times when needed.
* Given the memory requirements for computing one spectrum, you may want to calculate the spectrum for one isotopic species first, save the result into a file, and subsequently repeat the procedure for the second spectrum.
